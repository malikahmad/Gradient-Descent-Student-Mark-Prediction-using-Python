```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Inputs
study_hours = 5
previous_mark = 80
attendance = 90

# Initial weights
study_weight = 0.4
mark_weight = 0.5
attendance_weight = 0.2

# Actual value
actual_mark = 77

# Learning rate
learning_rate = 0.00001

# Training
for i in range(10):

    # Prediction
    prediction = (
        study_hours * study_weight +
        previous_mark * mark_weight +
        attendance * attendance_weight
    )

    # Error
    error = prediction - actual_mark

    # Loss
    loss = error ** 2

    # Gradients
    study_gradient = 2 * error * study_hours
    mark_gradient = 2 * error * previous_mark
    attendance_gradient = 2 * error * attendance

    # Update weights
    study_weight = study_weight - learning_rate * study_gradient
    mark_weight = mark_weight - learning_rate * mark_gradient
    attendance_weight = attendance_weight - learning_rate * attendance_gradient

    # New prediction
    new_prediction = (
        study_hours * study_weight +
        previous_mark * mark_weight +
        attendance * attendance_weight
    )

    print("\nIteration:", i + 1)
    print("Prediction:", prediction)
    print("Error:", error)
    print("Loss:", loss)
    print("Study Weight:", study_weight)
    print("Mark Weight:", mark_weight)
    print("Attendance Weight:", attendance_weight)
    print("New Prediction:", new_prediction)
```
